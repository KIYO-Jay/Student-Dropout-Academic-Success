# Student Dropout and Academic Success: Data Understanding & EDA

This notebook explores the official UCI dataset. Run `python ../download_dataset.py` from the project root first.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
DATA_PATH = ROOT / 'data' / 'student_dropout.csv'
df = pd.read_csv(DATA_PATH)
df.head()
```

## Dataset overview
Check the number of observations, columns, data types and basic descriptive statistics before selecting preprocessing steps.

```python
print('Shape:', df.shape)
display(df.info())
display(df.describe(include='all').T)
```

## Target distribution
The target has three classes and the source describes a class imbalance. This matters when choosing stratified splitting and macro-averaged metrics.

```python
target_counts = df['Target'].value_counts()
display(target_counts.to_frame('Count'))
ax = target_counts.plot(kind='bar', title='Student outcome distribution', rot=0)
ax.set_xlabel('Outcome')
ax.set_ylabel('Number of records')
plt.tight_layout()
plt.show()
```

## Data quality checks
The raw dataset should be checked locally rather than assuming there are no missing values or duplicates.

```python
quality = pd.DataFrame({
    'missing_values': df.isna().sum(),
    'missing_percent': (df.isna().mean() * 100).round(2),
    'unique_values': df.nunique(),
})
display(quality.sort_values('missing_values', ascending=False).head(15))
print('Duplicate rows:', df.duplicated().sum())
```

## Numeric feature distributions
Use histograms to inspect continuous/count variables. Coded nominal variables should not be interpreted as continuous measurements.

```python
numeric_cols = df.drop(columns=['Target']).select_dtypes(include='number').columns.tolist()
# A small selection of meaningful continuous/count measures for readable plots
selected = [c for c in numeric_cols if any(k in c.lower() for k in ['grade', 'age', 'unemployment', 'inflation', 'gdp', 'curricular units'])][:12]
df[selected].hist(figsize=(14, 10), bins=20)
plt.suptitle('Selected numeric feature distributions')
plt.tight_layout()
plt.show()
```

## Academic performance by outcome
These plots help examine how semester performance relates to the recorded outcome. Note that second-semester fields are only available after that semester, so this is not an admission-day prediction.

```python
performance_cols = [c for c in df.columns if c in [
    'Curricular units 1st sem (approved)', 'Curricular units 1st sem (grade)',
    'Curricular units 2nd sem (approved)', 'Curricular units 2nd sem (grade)'
]]
for col in performance_cols:
    ax = df.boxplot(column=col, by='Target', figsize=(7, 4), grid=False)
    plt.title(f'{col} by outcome')
    plt.suptitle('')
    plt.xlabel('Recorded outcome')
    plt.tight_layout()
    plt.show()
```
